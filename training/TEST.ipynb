```python
import numpy as np
counts = np.zeros_like(label_list, dtype="int64")
for dataname in ["test","eval","train"]:
    for lab_set in datasets[dataname]['label']:
        vals, cts = np.unique(lab_set, return_counts=True)
        for v, c in zip(vals, cts):
            counts[label2id[v]] += c
```
