```python
from Reader import obtain_combined_dataset
datasets, label_list, label2id, id2label = obtain_combined_dataset(["TempEval3"], "TempRel")
```

```python
label2id
```

```python
import numpy as np
values, counts = np.unique(datasets["train"]['label'], return_counts=True)
```

```python
from transformers import RobertaTokenizerFast, RobertaForSequenceClassification
tokenizer = RobertaTokenizerFast.from_pretrained("results\\TempRelTest")
model = RobertaForSequenceClassification.from_pretrained("results\\TempRelTest\\checkpoint-15000")
```

```python
top = 1000
```

```python
test_set = datasets["test"]["tokens"][0:]
```

```python
import torch
from sklearn.metrics import f1_score, precision_score, recall_score
encodings=tokenizer(test_set, padding=True, truncation=True, return_tensors="pt", is_split_into_words=True)
model.eval()
with torch.no_grad():
    outputs = model(**encodings)
    logits = outputs.logits
    predictions = torch.argmax(logits, dim=-1)
```

```python
preds = [id2label[pred.item()] for pred in list(predictions)]
ids = datasets["test"]["label"][0:]
```

```python
import numpy as np
def compute_metrics(predictions, ids):
    return {
        "precision": precision_score(ids, predictions, average="micro"),
        "recall": recall_score(ids, predictions, average="micro"),
        "f1": f1_score(ids, predictions, average="micro"),
    }

compute_metrics(preds, ids)
```

```python
from sklearn.metrics import classification_report
print(classification_report(ids, preds))
```
