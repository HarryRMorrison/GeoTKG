```python
import json

with open("test.json", "r") as f:
    examples=[json.loads(line) for line in f]
breaker = False
for example in examples:
    times = {}
    ets = {}
    for inst in example['instances']:
        if inst['type'] != "EVENT":
            times[inst['id']] = inst

    for et in example['event_times']:
        if et['event'] in ets:
            if times[et['time']]['type'] == "DURATION":
                breaker = True
            ets[et['event']].append(et['time'])
        else:
            ets[et['event']] = [et['time']]
    
    if breaker:
        break
```
